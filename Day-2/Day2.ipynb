```python
# frequency count
numbers = [10, 20, 10, 30, 20, 10]
unique=[]
count=0
for i in numbers:
  if i not in unique:
    unique.append(i)
for i in unique:
  count=0
  for j in numbers:
    if i==j:
      count+=1
  print(i,"appers",count,"times")


```

```python
# zeroes to the end
numbers = [0, 1, 0, 3, 12, 0, 5]
result=[]
zero_count=0
for i in numbers:
  if i==0:
    zero_count+=1
  else:
    result.append(i)
for i in range(zero_count):
  result.append(0)
print(result)

```

```python
# find the missing number in the sequence
numbers = [1, 2, 3, 5, 6]
n=len(numbers)+1
expected_sum=n*(n+1)//2
actual_sum=sum(numbers)
difference=expected_sum-actual_sum
print("Missing number:",difference)
```

```python
# find the missing nnumber in the sequence
numbers = [1, 2, 3, 5, 6, 7]
for i in range(len(numbers)-1):
  if (numbers[i+1])-(numbers[i])>1:
    missing=numbers[i]+1
print("missing number:",missing)
```

```python
# find the duplicates in the array
numbers = [10, 20, 30, 10, 40, 20, 50, 30]
unique=[]
duplicates=[]
for i in numbers:
  if i in unique:
    if i not in duplicates:
      duplicates.append(i)
  else:
    unique.append(i)
print("Duplicates numbers:",duplicates)

```

```python
# common elemnets in two arrays
A = [10, 20, 30, 40, 50]
B = [30, 40, 50, 60, 70]
common=[]
for i in A:
  if i in B:
    common.append(i)
print(common)
```

```python
# find the numbers in the given sum
numbers = [2, 4, 3, 5, 7, 8, 1]
target=9
for i in range(len(numbers)):
  for j in range(i+1,len(numbers)):
    if numbers[i]+numbers[j]==target:
      print(numbers[i],numbers[j])
```

```python
# Right rotation
numbers = [1, 2, 3, 4, 5]
k=2
for i in range(k):
  last=numbers.pop()
  numbers.insert(0,last)
print(numbers)
```

```python
# Left Rotation
numbers=[1,2,3,4,5]
k=2
for i in range(k):
  first=numbers.pop(0)
  numbers.append(first)
print(numbers)
```

```python
numbers = [2, 7, 1, 15, 3, 10]
smallest=numbers[0]
largest=numbers[0]
for i in numbers:
  if i<smallest:
    smallest=i
  if i> largest:
    largest=i
difference=largest-smallest
print("Largest differnce is",difference )
```

```python
# Move All Negative Numbers to the Beginning
numbers = [10, -5, 20, -8, 30, -2, 15]
positive=[]
negative=[]
for i in numbers:
  if i <0:
    negative.append(i)
  else:
    positive.append(i)
result=negative+positive
print(result)
```

```python
#Find the Longest Consecutive Sequence
numbers=[100,4,200,1,3,2]
number_set=set(numbers)
longest=0
for i in number_set:
  if i-1 not in number_set:
    current=i
    count=0
    while current in number_set:
      count+=1
      current+=1
    if count>longest:
      longest=count
print("Longest consecutive sequence length",longest)
```
