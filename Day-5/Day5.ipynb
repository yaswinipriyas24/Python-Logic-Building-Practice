```python
# Find the Largest of Three Numbers Using a Function
def find_largest(a,b,c):
  if a>=b and a>=c:
    return a
  elif b>=a and b>=c:
    return b
  else:
    return c
numbers=[]
for i in range(3):
  num=int(input(f"Enter Number{i+1}: "))
  numbers.append(num)
a,b,c=numbers
result=find_largest(a,b,c)
print(f"{result}is a biggest number")
```

```python
# Find the Factorial of a Number
def factorial(n):
  if n<0:
    return None
  elif n==0 or n==1 :
    return 1
  else:
    fact=1
    for i in range(2,n+1):
      fact=fact*i
    return fact
n=int(input("Enter a number"))
result=factorial(n)
if result is None:
  print("Factorial not defined for negative numbers")
else:
  print(f"Factorial {n} is {result}")
```

```python
# Check Prime Number Using a Function
def check_prime(n):
  if n<2:
    return False
  for i in range(2,n):
    if n%i==0:
      return False
  return True
n=int(input("enter a number"))
result=check_prime(n)
if result:
  print(f"{n} is a prime number")
else:
  print(f"{n} is not a prime number")
```

```python
# Check Palindrome Using a Function
def palindrome (n):
  if n==n[::-1]:
    return True
  else:
    return False
n=input("Enter a string")
result=palindrome(n)
print(result)
```

```python
# Check Palindrome Using a Function
def palindrome(n):
  reverse=""
  n=n.lower()
  for i in n:
    reverse=i+reverse
  if n==reverse:
    return True
  return False
n=input("Enter a string:")
result=palindrome(n)
if result:
  print("Palindrome")
else:
  print("Not Palindrome")

```

```python
# Count Vowels in a String Using a Function
def count_vowels(str):
  vowels="aeiouAEIOU"
  count=0
  for i in str:
    if i in vowels:
      count+=1
  return count
str=input("Enter a string:")
result=count_vowels(str)
print(f"Count of vowles in {str} are {result}")
```

```python
#Find the Second Largest Element Using a Function
def find_second_largest(numbers):
  max=float('-inf')
  sec_max=float('-inf')
  for i in numbers:
    if i>max:
      sec_max=max
      max=i
    else:
      if i>sec_max and i!=max:
        sec_max=i
  return sec_max
numbers=[10, 20, 40, 30, 50]
result=find_second_largest(numbers)
print("Second Largest number:",result)
```

```python
# Find Frequency of Each Element Using a Function
def frequency_count(numbers):
  frequency={}
  for i in numbers:
    if i in frequency:
      frequency[i]+=1
    else:
      frequency[i]=1
  return frequency

numbers=[1, 2, 2, 3, 3, 3, 4]
result=frequency_count(numbers)
print("Frequency count:",result)
```
