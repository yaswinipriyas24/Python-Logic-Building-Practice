```python
# Check Whether a List Is Sorted
def is_sorted(numbers):
  for i in range(len(numbers)-1):
    if numbers[i]>numbers[i+1]:
      return False
  return True
numbers=[1, 2, 2, 5, 8]
result=is_sorted(numbers)
if result:
  print("List is sorted in ascending order")
else:
  print("List is not sorted in ascending order")
```

```python
# Find the Third Largest Distinct Element
def third_largest(numbers):
  max=float('-inf')
  sec_max=float('-inf')
  third_max=float('-inf')
  for i in numbers:
    if i > max:
      third_max=sec_max
      sec_max=max
      max=i
    elif i>sec_max and i!=max:
      third_max=sec_max
      sec_max=i
    else:
      if i>third_max and i!=max and i!=sec_max:
        third_max=i
  return third_max
numbers=[10, 40, 20, 50, 30, 50]
third_max=third_largest(numbers)
print("Third Largest Distinct number:",third_max)
```

```python
#  Remove All Occurrences of a Given Element
def remove_element(numbers, target):
    result = []

    for i in numbers:
        if i != target:
            result.append(i)

    return result


numbers = [1, 2, 3, 2, 4, 2]
target = 2

result = remove_element(numbers, target)
print(result)
```

```python
# Merge Two Lists Alternately
def merge_lists(list1, list2):
    result = []

    for i in range(len(list1)):
        result.append(list1[i])
        result.append(list2[i])

    return result


list1 = [1, 3, 5]
list2 = [2, 4, 6]

result = merge_lists(list1, list2)
print(result)
```

```python
# Count Different Types of Characters
def count_characters(text):
    uppercase = 0
    lowercase = 0
    digits = 0
    spaces = 0
    special = 0

    for i in text:
        if i.isupper():
            uppercase += 1
        elif i.islower():
            lowercase += 1
        elif i.isdigit():
            digits += 1
        elif i.isspace():
            spaces += 1
        else:
            special += 1

    return uppercase, lowercase, digits, spaces, special


text = "Python 3.12 Is Fun!"

uppercase, lowercase, digits, spaces, special = count_characters(text)

print("Uppercase:", uppercase)
print("Lowercase:", lowercase)
print("Digits:", digits)
print("Spaces:", spaces)
print("Special characters:", special)
```

```python
# Reverse Each Word in a Sentence
def reverse_words(text):
    words = text.split()
    result = []

    for i in words:
        reverse = ""

        for j in i:
            reverse = j + reverse

        result.append(reverse)

    return " ".join(result)


text = "Python is easy"
result = reverse_words(text)

print("Reversed words:", result)
```

```python
# Check Whether Two Strings Are Rotations
def are_rotations(str1, str2):
    if len(str1) != len(str2):
        return False

    combined = str1 + str1

    if str2 in combined:
        return True
    else:
        return False


str1 = "abcd"
str2 = "cdab"

result = are_rotations(str1, str2)

if result:
    print("Strings are rotations")
else:
    print("Strings are not rotations")
```

```python
# Find the Most Frequent Word in a Sentence
def most_frequent_word(text):
    words = text.split()
    frequency = {}

    # Count word frequency
    for i in words:
        if i in frequency:
            frequency[i] += 1
        else:
            frequency[i] = 1

    # Find the most frequent word
    max_count = 0
    max_word = ""

    for i in frequency:
        if frequency[i] > max_count:
            max_count = frequency[i]
            max_word = i

    return max_word, max_count


text = "python is easy and python is powerful"

word, count = most_frequent_word(text)

print(word, "appears", count, "times")
```
