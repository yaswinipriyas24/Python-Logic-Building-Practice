```python
# Convert a nested list (a list containing multiple lists) into a single list.
def flatten_list(numbers):
  result=[]
  for i in numbers:
    for j in i:
      result.append(j)
  return result
numbers=[[1, 2], [3, 4], [5, 6]]
result=flatten_list(numbers)
print(result)
```

```python
#Find the Sum of Each Inner List
def inner_sum(numbers):
  result=[]
  for i in numbers:
    sum=0
    for j in i:
      sum+=j
    result.append(sum)
  return result
numbers= [[1, 2], [3, 4], [5, 6]]
result=inner_sum(numbers)
print(result)
```

```python
def sum_matrix(matrix):
    rows_sum = []
    col_sum = []

    # Calculate row sums
    for row in matrix:
        total = 0
        for value in row:
            total += value
        rows_sum.append(total)

    # Calculate column sums
    for col in range(len(matrix[0])):
        total = 0
        for row in range(len(matrix)):
            total += matrix[row][col]
        col_sum.append(total)

    return rows_sum, col_sum


matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

row_sum, col_sum = sum_matrix(matrix)

print("Row sums:", row_sum)
print("Column sums:", col_sum)
```

```python
# Transpose a Matrix
def transpose(matrix):
  result=[]
  for col in range(len(matrix[0])):
    new_row=[]
    for row in range(len(matrix)):
      new_row.append(matrix[row][col])
    result.append(new_row)
  return result
matrix = [
    [1, 2, 3],
    [4, 5, 6]
]
result=transpose(matrix)
print(result)
```

```python
# Find the Diagonal Sum of a Matrix
def diagnoal_sum(matrix):
  total=0
  for i in range(len(matrix)):
    for j in range(len(matrix)):
      if i==j:
        total+=matrix[i][j]
  return total
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
result=diagnoal_sum(matrix)
print(result)
```

```python
# Find Common Elements in Three Lists
def common_elements(list1,list2,list3):
  result=[]
  for i in list1:
    for j in list2:
      for k in list3:
        if i==j==k:
          result.append(i)
  return result
A = [1, 2, 3, 4, 5]
B = [2, 3, 5, 6]
C = [3, 5, 7, 8]
result=common_elements(A,B,C)
print(result)
```

```python
#Longest Substring Without Repeating Characters
def longest_substring(text):

    char_set = set()
    left = 0
    max_length = 0

    for right in range(len(text)):

        while text[right] in char_set:
            char_set.remove(text[left])
            left += 1

        char_set.add(text[right])

        current_length = right - left + 1

        max_length = max(max_length, current_length)

    return max_length


text = "abcabcbb"

result = longest_substring(text)

print("Longest substring length:", result)
```

```python
#Group words that contain the same characters, even when their character order differs.
def group_anagrams(words):
    groups = {}

    for word in words:
        key = "".join(sorted(word))

        if key in groups:
            groups[key].append(word)
        else:
            groups[key] = [word]

    return list(groups.values())


words = ["eat", "tea", "tan", "ate", "nat", "bat"]

result = group_anagrams(words)

print(result)
```

```python
# Check Whether Parentheses Are Balanced
def is_balanced(text):
    stack = []

    pairs = {
        ')': '(',
        '}': '{',
        ']': '['
    }

    for i in text:

        # Check opening brackets
        if i in "({[":
            stack.append(i)

        # Check closing brackets
        elif i in ")}]":
            if not stack:
                return False

            if stack[-1] == pairs[i]:
                stack.pop()
            else:
                return False

    return len(stack) == 0


text = "{[()]}"

result = is_balanced(text)

if result:
    print("Balanced")
else:
    print("Not Balanced")
```
