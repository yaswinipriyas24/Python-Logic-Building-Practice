```python
list=[]
n=int(input("enter no of numbers"))
for i in range(n):
  l=int(input(f"enter {i+1} numbers"))
  list.append(l)
print(list)
odd_count=0
even_count=0
for i in list:
  if i%2==0:
    even_count+=1
  else:
    odd_count+=1
print("odd count:",odd_count)
print("even count",even_count)
```
