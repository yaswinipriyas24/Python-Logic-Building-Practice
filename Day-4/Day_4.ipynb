```python
#Create and Access a Dictionary
employee={
    "name":"Rahul",
    "age":25,
    "department":"IT",
    "salary":30000
}
print(employee["name"])
print(employee["age"])
print(employee["department"])
print(employee["salary"])
```

```python
# Add,Update and Delete Dictionary Elements
products={
    "laptop":50000,
    "mobile":20000,
    "tablet":15000
}
products["keyboard"]=20000
products["mobile"]=25000
del products["tablet"]
print(products)
```

```python
# Frequency Count Using a Dictionary
numbers = [10, 20, 10, 30, 20, 10]
frequency={}
for i in numbers:
  if i in frequency:
    frequency[i]+=1
  else:
    frequency[i]=1
print(frequency)
```

```python
# Character Frequency Using a Dictionary
text="banana"
frequency={}
for i in text:
  if i in frequency:
    frequency[i]+=1
  else:
    frequency[i]=1
print(frequency)
```

```python
sentence = "python is easy and python is interesting"
words=sentence.split()
frequency={}
for i in words:
  if i in frequency:
    frequency[i]+=1
  else:
    frequency[i]=1
print(frequency)
```

```python
# Find the Key With the Maximum Value
marks= {
    "A":85,
    "B":92,
    "C":78,
    "D":95
}
max_marks=0
top_student=""
for i in marks:
  if marks[i]>max_marks:
    max_marks=marks[i]
    top_student=i
print("Top student name:",top_student)

print("Higest marks scored:",max_marks)
```

```python
# Merge Two Dictionaries
dict1={"a":10,"b":20,"d":30}
dict2={"b":30,"c":40}
result={}
for i in dict1:
  if i in dict2:
    result[i]=dict1[i]+dict2[i]
  else:
    result[i]=dict1[i]
for i in dict2:
  if i not in dict1:
    result[i]=dict2[i]
print(result)


```

```python
#Check Whether Two Strings Are Anagrams
str1="listen"
str2="silent"
freq1={}
freq2={}
for i in str1:
  if i in freq1:
    freq1[i]+=1
  else:
    freq1[i]=1
for i in str2:
  if i in freq2:
    freq2[i]+=1
  else:
    freq2[i]=1
if freq1==freq2:
  print("anagram")
else:
  print("not anagram")
```

```python
#Find the First Repeated Element
numbers=[10,20,30,40,20,50,10]
result={}
for i in numbers:
  if i in result:
    print(i,"first repeating digit")
    break
  else:
    result[i]=True
```

```python
# Two Sum Using a Dictionary
numbers = [2, 7, 11, 15]
target = 9
seen={}
for i in numbers:
  needed=target-i
  if needed in seen:
    print("Pair found:",needed,i)
    break
  else:
    seen[i]=True
```
