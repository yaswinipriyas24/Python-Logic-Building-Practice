```python
# Reverse an Array In Place using Two Pointers.
def reverse_array_in_place(arr):
  left=0
  right=len(arr)-1
  while left<right:
    arr[left],arr[right]=arr[right],arr[left]
    left+=1
    right-=1
  return arr
numbers = [5,10,15,20]
result=reverse_array_in_place(numbers)
print(result)
```

```python
# Find the Equilibrium Index
def equilibrium_index(numbers):
  for i in range(len(numbers)):
    left_sum=0
    right_sum=0
    for j in range(i):
      left_sum+=numbers[j]
    for j in range(i+1,len(numbers)):
      right_sum+=numbers[j]
    if left_sum==right_sum:
      return i
  return -1
numbers = [-7, 1, 5, 2, -4, 3, 0]

result = equilibrium_index(numbers)
print(result)
```

```python
# Sort an Array of 0s, 1s and 2s
def sort_colors(arr):
  low=0
  mid=0
  high=len(arr)-1
  while mid<=high:
    if arr[mid]==0:
      arr[low],arr[mid]=arr[mid],arr[low]
      low+=1
      mid+=1
    elif arr[mid]==1:
      mid+=1
    else:
      arr[mid],arr[high]=arr[high],arr[mid]
      high-=1
  return arr
numbers=[2,0,1,2,1,0,1]
result=sort_colors(numbers)
print(result)
```

```python
# Find All Repeating and Non-Repeating Elements
def find_repeating(numbers):
  repeating=[]
  non_repeating=[]
  for i in numbers:
    if i  in non_repeating:
      non_repeating.remove(i)
      repeating.append(i)
    else:
      non_repeating.append(i)
  return repeating ,non_repeating
numbers = [1, 2, 3, 2, 4, 5, 1, 6]
rep,non_rep=find_repeating(numbers)
print(rep)
print(non_rep)
```

```python
# Find All Repeating and Non-Repeating Elements
def find_repeating(numbers):
    frequency = {}

    for i in numbers:
        if i not in frequency:
            frequency[i] = 1
        else:
            frequency[i] += 1

    repeating = []
    non_repeating = []

    for key, value in frequency.items():
        if value > 1:
            repeating.append(key)
        else:
            non_repeating.append(key)

    return repeating, non_repeating


numbers = [1, 2, 3, 2, 4, 5, 1, 6]

rep, non_rep = find_repeating(numbers)

print("Repeating:", rep)
print("Non-repeating:", non_rep)
```

```python
# Remove duplicates from an unsorted array
def remove_duplicates(numbers):
  unique=[]
  for i in numbers:
    if i not in unique:
      unique.append(i)
  return unique
numbers = [1, 2, 3, 2, 4, 5, 1, 6]
result=remove_duplicates(numbers)
print(result)
```

```python
# Remove duplicates from an sorted array
def remove_duplicates_sorted(numbers):
    if not numbers:
        return []

    slow = 0
    fast = 1

    while fast < len(numbers):

        if numbers[slow] != numbers[fast]:
            slow += 1
            numbers[slow] = numbers[fast]

        fast += 1

    return numbers[:slow + 1]


numbers = [1, 1, 2, 2, 3, 4, 4, 5]

result = remove_duplicates_sorted(numbers)
print(result)
```
