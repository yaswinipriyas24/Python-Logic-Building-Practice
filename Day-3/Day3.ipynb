```python
# Reverse a string without using slice operations
text = "python"
reverse = ""
for i in text:
  reverse=i+reverse
print(reverse)
```

```python
# check palindrome or not
text="madam"
reverse=""
for char in text:
  reverse=char+reverse
if text== reverse:
  print("Palindrome")
else:
  print("Not a palindrome")
```

```python
# Count Vowels and Consonants
text = "programming"
vowels=0
consonants=0
for i in text:
  if i.lower() in "aeiou":
    vowels+=1
  elif i.isalpha():
    consonants+=1
print("Vowels:",vowels)
print("COnsonants:",consonants)
```

```python
# Character Frequency
text = "banana"
text=list(text)
unique=[]
for i in text:
  if i not in unique:
    unique.append(i)
for i in unique:
  count=0
  for j in text:
    if i==j:
      count+=1
  print(i,"appears",count)
```

```python
# First Non-Repeating Character
text = "swiss"
text=list(text)
unique=[]
for i in text:
  if i not in unique:
    unique.append(i)
for i in unique:
  count=0
  for j in text:
    if i==j:
      count+=1
  if count==1:
    print("First repeating character:",i)
    break
```

```python
#Remove Duplicate Characters
text = "programming"
unique=[]
for i in text:
  if i not in unique:
    unique.append(i)
result="".join(unique)
print("Unique characters are:",result)
```

```python
# Find the Longest Word in a Sentence.
sentence = "Python programming is interesting"
words = sentence.split()
longest_word=""
for i in words:
  if len(i)>len(longest_word):
    longest_word=i
print("longest_word is:",longest_word)
```

```python
# Rotate a String by K Positions
text="python"
k=2
text=list(text)
for i in range (k):
  char=text.pop()
  text.insert(0,char)
print(text)
```

```python
# Rotate a String by K Positions
text="python"
k=2
result=text[-k:]+text[:-k]
print(result)
```

```python
# Find the Most Frequent Character
text="mississippi"
unique=[]
for i in text:
  if i not in unique:
    unique.append(i)
max_count=0
max_char=""
for i in unique:
  count=0
  for j in text:
    if i==j:
      count+=1
  if count>max_count:
    max_count=count
    max_char=[i]
  elif count==max_count:
    max_char.append(i)
print("Most frequent character is",max_char,"appears",max_count)
```
